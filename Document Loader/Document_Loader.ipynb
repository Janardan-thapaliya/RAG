**Text Loader**

```python
from pprint import pp
from langchain_community.document_loaders import TextLoader
from pathlib import Path
```

```python
file_path = Path("/content/transformers.txt")
print(file_path.exists())
```

```python
#define the loader
loader = TextLoader(file_path=file_path, encoding="latin-1")

#load the document
documents = loader.load()

print(type(documents))
```

```python
#Number of document
len(documents)
```

The number of document is 1 because Text Loaders load whole document as single object. It is different from pdfloaders which do it page wise.

##JSON Loader

```python
from langchain_community.document_loaders.json_loader import JSONLoader
from pathlib import Path
from pprint import pp
```

```python
# Create JSON path

json_path = Path('/content/apparels.json')

json_path.exists()
```

```python
#Create metadata function: Takes both dict and returns the updated dict

def func_metadata(record: dict, metadata: dict) -> dict:
  metadata['product_name'] = record['productName']
  metadata['category'] = record['category']
  metadata['price'] = record['price']

  return metadata
```

```python
#Form the loader

json_loader = JSONLoader(file_path=json_path,
                         jq_schema='.products[]', #Provide whole or some elements of schema
                         content_key='Description', #Choose which element will be content
                         metadata_func=func_metadata #Call function to add to metadata
)
```

```python
# Load the documents

json_doc = json_loader.load()
```

```python
for doc in json_doc:
  print(doc.page_content, end='\n\n')
```

```python
json_doc[0].metadata
```

```python
len(json_doc)
```

### Recursive loader

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
# Create loader

base_url = "https://www.ibm.com/events/think"

rc_web_loader = RecursiveUrlLoader(url=base_url,
                                   max_depth=2)
```

```python
# Load the documents

doc_web_rc = rc_web_loader.load()
```

```python
len(doc_web_rc)
```

```python
#Metadata for first 10 docs

for i in range(10):
  doc_w = doc_web_rc[i]
  print(doc_w.metadata)
```

```python
# Lazy load

documents_lazy_load = rc_web_loader.lazy_load()
```

```python
#Type

print(type(documents_lazy_load))
```

**Instead of waiting to load everything at once, the generator takes one item, loads it in memory and then removes it before loading another one**

### Works like streaming

```python
counter = 0

for document in documents_lazy_load:
  #Stop condition
  if counter == 10:
    break

  #increment counter
  counter += 1

  #print metadata and page content
  print(document.page_content[0:300])
  pp(document.metadata)
```
