```python
from dotenv import load_dotenv
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_openai import OpenAIEmbeddings
```

```python
# load env vars
load_dotenv()
```

```python
# models

MODEL_LARGE = "text-embedding-3-large"

MODEL_SMALL = "text-embedding-3-small"
```

```python
# create the embedder

embedder_large = OpenAIEmbeddings(
    model=MODEL_LARGE
)

embedder_small = OpenAIEmbeddings(
    model=MODEL_SMALL
)
```

```python
# create embeddings from user query

query = "What is Openclaw/Moltbot and what are the major security concerns regarding this tool"

embeddings_large = embedder_large.embed_query(
    text=query
)
```

```python
len(embeddings_large)
```

```python
embeddings_large[0:10]
```

```python
embeddings_small = embedder_small.embed_query(
    text=query
)

len(embeddings_small)
```

```python
embeddings_small[0:10]
```

```python
# create the loader

loader = PyPDFLoader(file_path="/content/Openclaw_Research_Report.pdf")

docs = loader.load()

len(docs) #Pages
```

```python
#Split documents

chunker = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
)

chunks = chunker.split_documents(
    documents=docs
)

len(chunks)
```

```python
# Embed documents

text_documents = [doc.page_content for doc in chunks]

print(len(text_documents))

document_embeddings = embedder_large.embed_documents(
    texts=text_documents
)
```

```python
len(document_embeddings)
```

```python
document_embeddings[0:3]
```

```python
# change the dimensions

embedder_large_custom_dim = OpenAIEmbeddings(
    model=MODEL_LARGE,
    dimensions=256
)
```

```python
query_embeddings = embedder_large_custom_dim.embed_query(text=query)
```

```python
len(query_embeddings)
```

```python
from dotenv import load_dotenv
import ollama
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_ollama.embeddings import OllamaEmbeddings
```

```python
query = "What is Openclaw/Moltbot and what are the major security concerns regarding this tool"
```

```python
# Native ollama embedding - If using langchain or langgraph, it can throw error so use OllamaEmbeddings which is Langchain integration

query_embeddings = ollama.embed(
    model="embeddinggemma",
    input=query
)
```

```python
query_embeddings["embeddings"]
```

```python
# dimensions

print(len(query_embeddings["embeddings"][0]))
```

```python
query_embeddings = ollama.embed(
    model="embeddinggemma",
    input=query,
    dimensions=512
)
```

```python
# dimensions

print(len(query_embeddings["embeddings"][0]))
```

```python
# create the loader

loader = PyPDFLoader(file_path="/content/Openclaw_Research_Report.pdf")

docs = loader.load()

len(docs)
```

```python
# split documents

chunker = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = chunker.split_documents(docs)

len(chunks)
```

```python
# embed documents

text_documents = [doc.page_content for doc in chunks]

print(len(text_documents))
```

```python
# embed documents

document_embeddings = ollama.embed(
    model="embeddinggemma",
    input=text_documents
)
```

```python
document_embeddings = document_embeddings["embeddings"]
```

```python
# number of embedding vectors

print(len(document_embeddings))
```

```python
# output dimensions

print(len(document_embeddings[0]))
```
